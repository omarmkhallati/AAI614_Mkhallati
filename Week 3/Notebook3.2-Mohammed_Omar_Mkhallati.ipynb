Question 2: There are various data that are missing.  Fill-in the missing data or delete the rows/columns that you think you should delete.  Justify your answer

The code fills in the two missing values rather than deleting any rows or columns. George is missing his GPA, and Frank and Rachel are missing their Sense of Humour score, that's only one gap per row out of six total columns, so each of those rows still holds five good, usable data points. Deleting those rows would mean throwing away all of that other information just to get rid of one missing cell, which is a poor trade-off, especially with a dataset this small (only 9 students to begin with). Instead, the code uses fillna() with the column mean for GPA and for Sense of Humour,

```python
grads['GPA'] = grads['GPA'].fillna(grads['GPA'].mean())
grads['Sense of Humour (0-5)'] = grads['Sense of Humour (0-5)'].fillna(grads['Sense of Humour (0-5)'].mean())

grads
```

Question 3: Reload the data and fill-in the data using mean method as well as the frequent method.

This code reloads the original dataset fresh (so it's unaffected by the fill already applied in Question 2), then creates two separate copies to compare imputation strategies. One copy fills the missing GPA and Sense of Humour values using the mean, the average of each column, which works well here since both are numeric scores that vary somewhat continuously. The other copy fills the same missing values using the mode, the most frequently occurring value in each column, which is typically better suited to categorical or repeated-value data rather than continuous scores.

```python
grads_reload = pd.read_csv("https://raw.githubusercontent.com/harmanani/AAI614/main/Week%203/grads.csv")

grads_mean = grads_reload.copy()
grads_mean['GPA'] = grads_mean['GPA'].fillna(grads_mean['GPA'].mean())
grads_mean['Sense of Humour (0-5)'] = grads_mean['Sense of Humour (0-5)'].fillna(grads_mean['Sense of Humour (0-5)'].mean())

grads_freq = grads_reload.copy()
grads_freq['GPA'] = grads_freq['GPA'].fillna(grads_freq['GPA'].mode()[0])
grads_freq['Sense of Humour (0-5)'] = grads_freq['Sense of Humour (0-5)'].fillna(grads_freq['Sense of Humour (0-5)'].mode()[0])

print("Mean method:")
print(grads_mean)
print("\nMost frequent (mode) method:")
print(grads_freq)
```

Exercise II. Run the cell below to create a new dataframe called `df_miss`.  Its first column will contain some missing values.

```python
import pandas as pd
import numpy as np
import random

nrows = 10
ncols = 5

# set a seed for random number generation
np.random.seed(314)
# create an array filled with random data
data = np.array(np.random.rand(nrows, ncols))
# put the data to a pandas dataframe
df_miss = pd.DataFrame(data)
# rename the columns
df_miss.columns = ['col_'+str(ii) for ii in range(ncols)]

# randomly set some values to missing
ix0 = np.random.randint(nrows, size=3)
ix1 = np.random.randint(nrows, size=3)

df_miss['col_0'][ix0] = np.nan
df_miss['col_1'][ix1] = np.nan

print(df_miss)
```

Impute the missing values (NaN) in `col_0` (but not `col_1`) with the median.  Store the values in the dataframe by using the parameter `inplace`.  Print the dataframe.

```python
df_miss['col_0'].fillna(df_miss['col_0'].median(), inplace=True)
print(df_miss)
```

Impute the missing values in `col_1` with value 0.  Store the values in the dataframe by using the parameter `inplace`.  Print the dataframe.

```python
df_miss['col_1'].fillna(0, inplace=True)
print(df_miss)
```
